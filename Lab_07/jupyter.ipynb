#   Problem 1: Exam Scheduling

```python
# --- Variables & Domains ---
subjects  = ["Math", "Physics", "Chemistry"]
time_slots = ["T1", "T2"]

# Conflicts: pairs that CANNOT share the same time slot
conflicts = [
    ("Math",    "Physics"),
    ("Physics", "Chemistry"),
]

def is_consistent_exam(assignment, subject, slot):
    """Check if assigning 'slot' to 'subject' violates any constraint."""
    for (s1, s2) in conflicts:
        if s1 == subject and s2 in assignment:
            if assignment[s2] == slot:
                return False
        if s2 == subject and s1 in assignment:
            if assignment[s1] == slot:
                return False
    return True

def backtrack_exam(assignment, subjects, time_slots):
    """Backtracking search for exam scheduling."""
    if len(assignment) == len(subjects):
        return assignment                          # all subjects assigned

    subject = subjects[len(assignment)]            # pick next unassigned subject

    for slot in time_slots:
        if is_consistent_exam(assignment, subject, slot):
            assignment[subject] = slot
            result = backtrack_exam(assignment, subjects, time_slots)
            if result:
                return result
            del assignment[subject]                # backtrack

    return None

solution1 = backtrack_exam({}, subjects, time_slots)

print("\n Variables  :", subjects)
print("Domains    :", time_slots)
print("Constraints: Math≠Physics slot | Physics≠Chemistry slot")

if solution1:
    print("\n Valid Schedule Found:")
    print("-" * 30)
    for subject, slot in solution1.items():
        print(f"   {subject:<12} → {slot}")
    print("-" * 30)
    print("\n Verification:")
    for (s1, s2) in conflicts:
        clash = solution1[s1] == solution1[s2]
        status = " CLASH" if clash else "OK"
        print(f"   {s1} ({solution1[s1]}) vs {s2} ({solution1[s2]}) → {status}")
else:
    print("\n No valid schedule found.")
```

# Problem 2: House Assignment

```python
# --- Variables & Domains ---
people = ["A", "B", "C"]
houses = ["H1", "H2", "H3"]

# Restrictions: (person, forbidden_house)
restrictions = [
    ("A", "H1"),
    ("B", "H2"),
]

def is_consistent_house(assignment, person, house):
    """Check if assigning 'house' to 'person' violates any constraint."""
    # No two people share the same house
    if house in assignment.values():
        return False
    # Person-specific restrictions
    for (p, forbidden) in restrictions:
        if p == person and house == forbidden:
            return False
    return True

def backtrack_house(assignment, people, houses):
    """Backtracking search for house assignment."""
    if len(assignment) == len(people):
        return assignment                          # all people assigned

    person = people[len(assignment)]               # pick next unassigned person

    for house in houses:
        if is_consistent_house(assignment, person, house):
            assignment[person] = house
            result = backtrack_house(assignment, people, houses)
            if result:
                return result
            del assignment[person]                 # backtrack

    return None

solution2 = backtrack_house({}, people, houses)

print("\n Variables  :", people)
print(" Domains    :", houses)
print(" Constraints: A≠H1 | B≠H2 | all houses unique")

if solution2:
    print("\n Valid House Assignment Found:")
    print("-" * 30)
    for person, house in solution2.items():
        print(f"   Person {person}  →  {house}")
    print("-" * 30)
    print("\n Verification:")
    for (p, forbidden) in restrictions:
        assigned = solution2[p]
        ok = assigned != forbidden
        print(f"   Person {p} assigned {assigned} (forbidden={forbidden}) → {' OK' if ok else '❌ VIOLATION'}")
    houses_used = list(solution2.values())
    unique = len(houses_used) == len(set(houses_used))
    print(f"   All houses unique → {' OK' if unique else '❌ VIOLATION'}")
else:
    print("\n No valid assignment found.")
```

```python
# All exam schedules
def find_all_exam_solutions():
    solutions = []
    def bt(assignment):
        if len(assignment) == len(subjects):
            solutions.append(dict(assignment))
            return
        subject = subjects[len(assignment)]
        for slot in time_slots:
            if is_consistent_exam(assignment, subject, slot):
                assignment[subject] = slot
                bt(assignment)
                del assignment[subject]
    bt({})
    return solutions

# All house assignments
def find_all_house_solutions():
    solutions = []
    def bt(assignment):
        if len(assignment) == len(people):
            solutions.append(dict(assignment))
            return
        person = people[len(assignment)]
        for house in houses:
            if is_consistent_house(assignment, person, house):
                assignment[person] = house
                bt(assignment)
                del assignment[person]
    bt({})
    return solutions

all_exam   = find_all_exam_solutions()
all_house  = find_all_house_solutions()

print(f"\n Exam Scheduling  — {len(all_exam)} valid solution(s):")
for i, sol in enumerate(all_exam, 1):
    print(f"   Solution {i}: {sol}")

print(f"\n House Assignment — {len(all_house)} valid solution(s):")
for i, sol in enumerate(all_house, 1):
    print(f"   Solution {i}: {sol}")
```
